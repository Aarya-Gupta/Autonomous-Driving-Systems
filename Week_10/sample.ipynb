```python
import numpy as np
import scipy.stats as stats

# Number of samples
num_samples = 15

# Generate samples from different distributions
normal_samples = stats.norm.rvs(loc=0, scale=1, size=num_samples)       # Normal distribution with mean=0 and std=1
uniform_samples = stats.uniform.rvs(loc=0, scale=1, size=num_samples)   # Uniform distribution between 0 and 1
exponential_samples = stats.expon.rvs(scale=1, size=num_samples)        # Exponential distribution with scale=1
beta_samples = stats.beta.rvs(a=5, b=1, size=num_samples)               # Beta distribution with a=2, b=5
gamma_samples = stats.gamma.rvs(a=2, scale=1, size=num_samples)         # Gamma distribution with shape=2 and scale=1

# Combine all samples into a dictionary for easy access
samples = {
    "Normal": normal_samples,
    "Uniform": uniform_samples,
    "Exponential": exponential_samples,
    "Beta": beta_samples,
    "Gamma": gamma_samples
}

# Print the samples for each distribution
for dist_name, dist_samples in samples.items():
    print(f"{dist_name} samples:\n{dist_samples}\n")
```
